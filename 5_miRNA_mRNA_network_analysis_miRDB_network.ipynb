6. Define row-wise standardization and map measured miRNA to TargetSeed 8.0

```python
def zscore_rows(df):
    row_means = df.mean(axis=1)
    row_sd = df.std(axis=1, ddof=1)

    z = df.sub(row_means, axis=0).div(row_sd, axis=0)

    # Remove a feature only if it has no variation
    return z.dropna(axis=0, how="any")
```

```python
seed_map = (
    mirna_annotation
    .drop_duplicates("miRNA_original")
    .set_index("miRNA_original")["Seed+m8"]
)

mirna_seed_mapping = pd.DataFrame(
    {
        "miRNA": mirna_matched.index,
        "Seed": seed_map.reindex(mirna_matched.index).values
    }
)

mapped_mask = mirna_seed_mapping["Seed"].notna()

print("Measured miRNAs:", len(mirna_seed_mapping))
print("Mapped miRNAs:", mapped_mask.sum())
print("Unmapped miRNAs:", (~mapped_mask).sum())

display(mirna_seed_mapping.loc[~mapped_mask])
```

7.) Standardize the miRNAs and aggregate them by seed

```python
mapped_mirnas = mirna_seed_mapping.loc[
    mapped_mask, "miRNA"
].tolist()

mirna_z = zscore_rows(
    mirna_matched.loc[mapped_mirnas]
)

# Add the seed corresponding to each standardized miRNA
mirna_z["Seed"] = seed_map.loc[mirna_z.index]

# Average standardized expression across miRNAs sharing a seed
seed_z = (
    mirna_z
    .groupby("Seed")
    .mean(numeric_only=True)
)

print("Seed-expression matrix:", seed_z.shape)
```

```python
#Standardize the gene-expression matrix:
gene_z = zscore_rows(mrna_matched)

print("Gene-expression matrix:", gene_z.shape)
```

```python
#verify column match
assert seed_z.columns.equals(gene_z.columns)

print("Profiles:", len(seed_z.columns))
print("Network edges:", len(network))
```

```python
#extract seed scores
seed_edge_z = seed_z.loc[
    network["Seed"]
].to_numpy()

gene_edge_z = gene_z.loc[
    network["Gene"]
].to_numpy()

#calculate difference
edge_contrast = seed_edge_z - gene_edge_z

#apply logistic transformation
edge_score_values = 1 / (1 + np.exp(-edge_contrast))

#store the result as a labeled DataFrame:
edge_index = pd.MultiIndex.from_frame(
    network[["Seed", "Gene"]]
)

edge_scores = pd.DataFrame(
    edge_score_values,
    index=edge_index,
    columns=seed_z.columns
)

edge_scores.index.names = ["Seed", "Gene"]

print("Edge-score matrix:", edge_scores.shape)
display(edge_scores.head())

#check the range
print(
    "Minimum edge score:",
    edge_scores.to_numpy().min()
)

print(
    "Maximum edge score:",
    edge_scores.to_numpy().max()
)

print(
    "Mean edge score:",
    edge_scores.to_numpy().mean()
)
```

11.) Calculate the six network features. 

```python
#define normalized entropy
def normalized_entropy(values):
    values = np.asarray(values, dtype=float)

    probabilities = values / values.sum()

    entropy = -np.sum(
        probabilities * np.log(probabilities)
    )

    maximum_entropy = np.log(len(values))

    return entropy / maximum_entropy
```

Calculate the features for each matched profile

```python
feature_rows = []

for profile_id in edge_scores.columns:

    # All edge scores for this profile
    scores = edge_scores[profile_id]

    # Mean activity of each seed family across its target edges
    family_outstrength = (
        scores
        .groupby(level="Seed")
        .mean()
    )

    # Mean activity of each target gene across its regulating seeds
    target_instrength = (
        scores
        .groupby(level="Gene")
        .mean()
    )

    feature_rows.append(
        {
            "profile_id": profile_id,

            # 1. Mean score over all network edges
            "mean_repression_score": scores.mean(),

            # 2. Fraction of edges with score >= 0.75
            "active_edge_fraction": (
                scores >= 0.75
            ).mean(),

            # 3. Dispersion of individual edge scores
            "edge_score_dispersion": scores.std(ddof=1),

            # 4. Variation in mean activity among seed families
            "family_outstrength_cv": (
                family_outstrength.std(ddof=1)
                / family_outstrength.mean()
            ),

            # 5. Variation in mean incoming activity among targets
            "target_instrength_cv": (
                target_instrength.std(ddof=1)
                / target_instrength.mean()
            ),

            # 6. Evenness of activity across seed families
            "family_outstrength_entropy": normalized_entropy(
                family_outstrength
            ),
        }
    )

network_features = pd.DataFrame(feature_rows)

#show features
network_features
```

14.) Freedman Lane permutation test

```python
#import package
from scipy.stats import t as student_t
```

```python
#define the OLS fitting function:
def ols_fit(y, X):

    beta, _, _, _ = np.linalg.lstsq(
        X,
        y,
        rcond=None
    )

    fitted = X @ beta
    residuals = y - fitted

    degrees_freedom = len(y) - X.shape[1]
    residual_sum_squares = residuals @ residuals
    residual_variance = (
        residual_sum_squares / degrees_freedom
    )

    covariance = (
        residual_variance
        * np.linalg.pinv(X.T @ X)
    )

    standard_errors = np.sqrt(
        np.diag(covariance)
    )

    t_values = beta / standard_errors

    p_values = 2 * student_t.sf(
        np.abs(t_values),
        degrees_freedom
    )

    total_sum_squares = np.sum(
        (y - y.mean()) ** 2
    )

    r_squared = (
        1 - residual_sum_squares / total_sum_squares
    )

    return {
        "beta": beta,
        "fitted": fitted,
        "residuals": residuals,
        "t": t_values,
        "p": p_values,
        "r_squared": r_squared,
    }
```

Define Freedman Lane interaction test

```python
def freedman_lane_interaction_test(
    y,
    species_indicator,
    age,
    n_permutations=5000,
    rng=None,
):

    if rng is None:
        rng = np.random.default_rng(29356)

    # Full model:
    # outcome ~ species + age + species:age
    X_full = np.column_stack(
        [
            np.ones(len(y)),
            species_indicator,
            age,
            species_indicator * age,
        ]
    )

    # Reduced null model:
    # outcome ~ species + age
    X_reduced = np.column_stack(
        [
            np.ones(len(y)),
            species_indicator,
            age,
        ]
    )

    full_fit = ols_fit(y, X_full)
    reduced_fit = ols_fit(y, X_reduced)

    observed_t = full_fit["t"][3]
    extreme_count = 0

    for _ in range(n_permutations):

        permuted_residuals = rng.permutation(
            reduced_fit["residuals"]
        )

        permuted_y = (
            reduced_fit["fitted"]
            + permuted_residuals
        )

        permuted_fit = ols_fit(
            permuted_y,
            X_full
        )

        permuted_t = permuted_fit["t"][3]

        if abs(permuted_t) >= abs(observed_t):
            extreme_count += 1

    permutation_p = (
        extreme_count + 1
    ) / (
        n_permutations + 1
    )

    return {
        "chimpanzee_slope": full_fit["beta"][2],
        "human_slope": (
            full_fit["beta"][2]
            + full_fit["beta"][3]
        ),
        "human_minus_chimpanzee_slope": (
            full_fit["beta"][3]
        ),
        "interaction_t": observed_t,
        "analytic_p": full_fit["p"][3],
        "permutation_p": permutation_p,
        "r_squared": full_fit["r_squared"],
        "n": len(y),
    }
```

Run test for all six features

```python
rng = np.random.default_rng(29356)

species_indicator = (
    human_chimp["Species"] == "human"
).astype(int).to_numpy()

aligned_age = human_chimp[
    "log2_macaque_equivalent_age_plus_1"
].to_numpy()

test_rows = []

for feature in feature_columns:

    print("Testing:", feature)

    y = human_chimp[feature].to_numpy()

    result = freedman_lane_interaction_test(
        y=y,
        species_indicator=species_indicator,
        age=aligned_age,
        n_permutations=5000,
        rng=rng,
    )

    result["feature"] = feature
    test_rows.append(result)

permutation_results = pd.DataFrame(test_rows)
```

Define Benjamini–Hochberg FDR correction

```python
def benjamini_hochberg(p_values):

    p_values = np.asarray(p_values)
    number_tests = len(p_values)

    order = np.argsort(p_values)
    ranked_p = p_values[order]

    adjusted = (
        ranked_p
        * number_tests
        / np.arange(1, number_tests + 1)
    )

    adjusted = np.minimum.accumulate(
        adjusted[::-1]
    )[::-1]

    adjusted = np.minimum(
        adjusted,
        1.0
    )

    result = np.empty(number_tests)
    result[order] = adjusted

    return result
```

```python
#apply correction
permutation_results["fdr_bh"] = (
    benjamini_hochberg(
        permutation_results["permutation_p"]
    )
)

permutation_results = (
    permutation_results
    .sort_values("permutation_p")
    .reset_index(drop=True)
)
```

16.) Network-specificity test with 5000 randomized target labels

This test asks whether the observed human–chimpanzee developmental interaction for
target_instrength_cv and edge_score_dispersion is specific to the biological
miRNA–target assignments.

The target-gene expression profiles are randomly relabeled among the existing target
nodes. This preserves the full network topology, including every seed out-degree and
target in-degree, while breaking the biological correspondence between a target node
and its measured mRNA profile. The original edge-score definition and the original
species × age regression are otherwise unchanged.


```python
# Network-specificity randomization test
# Existing notebook objects are reused unchanged.

from scipy.special import expit

N_NETWORK_RANDOMIZATIONS = 5000
network_rng = np.random.default_rng(29356)

# Use exactly the same human/chimpanzee profiles and age variable as above.
hc_profiles = human_chimp["profile_id"].tolist()

species_indicator_network = (
    human_chimp["Species"] == "human"
).astype(int).to_numpy()

aligned_age_network = human_chimp[
    "log2_macaque_equivalent_age_plus_1"
].to_numpy()

X_network = np.column_stack(
    [
        np.ones(len(human_chimp)),
        species_indicator_network,
        aligned_age_network,
        species_indicator_network * aligned_age_network,
    ]
)

X_network_pinv = np.linalg.pinv(X_network)
X_network_xtx_inv = np.linalg.pinv(X_network.T @ X_network)
network_df = len(human_chimp) - X_network.shape[1]

def fixed_design_interaction_t(y):
    """OLS t-statistic for the species × age coefficient."""
    y = np.asarray(y, dtype=float)
    beta = X_network_pinv @ y
    residuals = y - X_network @ beta
    residual_variance = (residuals @ residuals) / network_df
    interaction_se = np.sqrt(
        residual_variance * X_network_xtx_inv[3, 3]
    )
    return beta[3] / interaction_se

# Seed expression for every original edge, restricted to human/chimp profiles.
seed_edge_z_hc = (
    seed_z.loc[network["Seed"], hc_profiles]
    .to_numpy(dtype=float)
)

# Each target node keeps its exact position and degree in the network.
# Only the mRNA expression profile assigned to that node is permuted.
target_nodes = pd.Index(network["Gene"].drop_duplicates())

target_expression_hc = (
    gene_z.loc[target_nodes, hc_profiles]
    .to_numpy(dtype=float)
)

edge_target_index = target_nodes.get_indexer(network["Gene"])

# Precompute grouping information for fast target in-strength calculation.
edge_order_by_target = np.argsort(edge_target_index)
sorted_target_index = edge_target_index[edge_order_by_target]

target_starts = np.r_[
    0,
    np.flatnonzero(np.diff(sorted_target_index)) + 1
]

target_counts = np.diff(
    np.r_[target_starts, len(sorted_target_index)]
)

random_t_target_cv = np.empty(N_NETWORK_RANDOMIZATIONS)
random_t_edge_dispersion = np.empty(N_NETWORK_RANDOMIZATIONS)

for b in range(N_NETWORK_RANDOMIZATIONS):

    # One-to-one relabeling among the measured network targets.
    # This preserves the entire seed/target degree structure.
    target_permutation = network_rng.permutation(len(target_nodes))

    permuted_target_expression = target_expression_hc[
        target_permutation
    ]

    gene_edge_z_random = permuted_target_expression[
        edge_target_index
    ]

    random_edge_scores = expit(
        seed_edge_z_hc - gene_edge_z_random
    )

    # Same edge-score dispersion definition as in the original analysis.
    random_edge_dispersion = random_edge_scores.std(
        axis=0,
        ddof=1
    )

    # Same target in-strength definition as in the original analysis:
    # mean incoming edge score for each target, followed by CV across targets.
    sorted_scores = random_edge_scores[
        edge_order_by_target
    ]

    random_target_instrength = (
        np.add.reduceat(
            sorted_scores,
            target_starts,
            axis=0
        )
        / target_counts[:, None]
    )

    random_target_cv = (
        random_target_instrength.std(axis=0, ddof=1)
        / random_target_instrength.mean(axis=0)
    )

    random_t_target_cv[b] = fixed_design_interaction_t(
        random_target_cv
    )

    random_t_edge_dispersion[b] = fixed_design_interaction_t(
        random_edge_dispersion
    )

    if (b + 1) % 500 == 0:
        print(
            f"Completed {b + 1} / "
            f"{N_NETWORK_RANDOMIZATIONS} randomized networks"
        )

# Observed t-statistics from the unchanged original analysis.
observed_t = (
    permutation_results
    .set_index("feature")["interaction_t"]
)

def summarize_network_null(feature, random_t):
    obs = float(observed_t.loc[feature])

    # Percentile of the observed topology within the randomized-topology null.
    percentile = 100 * np.mean(random_t <= obs)

    # Two-sided empirical p-value based on the location of the observed
    # statistic within the randomized-network distribution.
    lower_tail = (
        1 + np.sum(random_t <= obs)
    ) / (len(random_t) + 1)

    upper_tail = (
        1 + np.sum(random_t >= obs)
    ) / (len(random_t) + 1)

    network_specific_p = min(
        1.0,
        2 * min(lower_tail, upper_tail)
    )

    return {
        "feature": feature,
        "observed_interaction_t": obs,
        "random_mean_t": random_t.mean(),
        "random_median_t": np.median(random_t),
        "random_sd_t": random_t.std(ddof=1),
        "observed_percentile": percentile,
        "network_specific_p": network_specific_p,
    }

network_specificity_results = pd.DataFrame(
    [
        summarize_network_null(
            "target_instrength_cv",
            random_t_target_cv
        ),
        summarize_network_null(
            "edge_score_dispersion",
            random_t_edge_dispersion
        ),
    ]
)

display(
    network_specificity_results.round(6)
)

network_specificity_results.to_csv(
    "results/pfc_network_specificity_5000_randomized_targets.csv",
    index=False
)

# Save the full randomized t distributions for transparency/reproducibility.
pd.DataFrame(
    {
        "randomization": np.arange(
            1,
            N_NETWORK_RANDOMIZATIONS + 1
        ),
        "target_instrength_cv_t": random_t_target_cv,
        "edge_score_dispersion_t": random_t_edge_dispersion,
    }
).to_csv(
    "results/pfc_network_specificity_5000_randomized_targets_full.csv",
    index=False
)
```

17.) Check whether the original network metrics track expression heterogeneity

Reviewer's criticism predicts that the network summaries may largely reflect the
cross-feature spread of seed-family and mRNA expression within each profile. We therefore
calculate simple SD and MAD measures from the same standardized seed and gene matrices and
correlate them with the two network metrics. This is a diagnostic analysis; it does not alter
the original network analysis.


We check for expression heterogenity

```python
# Simple profile-level expression heterogeneity diagnostics

from scipy.stats import pearsonr

def column_mad(df):
    values = df.to_numpy(dtype=float)
    medians = np.median(values, axis=0)
    return np.median(
        np.abs(values - medians),
        axis=0
    )

expression_heterogeneity = pd.DataFrame(
    {
        "profile_id": seed_z.columns,
        "mirna_seed_sd": seed_z.std(axis=0, ddof=1).to_numpy(),
        "mirna_seed_mad": column_mad(seed_z),
        "mrna_sd": gene_z.std(axis=0, ddof=1).to_numpy(),
        "mrna_mad": column_mad(gene_z),
    }
)

expression_heterogeneity = expression_heterogeneity.merge(
    network_features[
        [
            "profile_id",
            "Species",
            "match_age_days",
            "target_instrength_cv",
            "edge_score_dispersion",
        ]
    ],
    on="profile_id",
    how="left",
    validate="one_to_one"
)

heterogeneity_columns = [
    "mirna_seed_sd",
    "mirna_seed_mad",
    "mrna_sd",
    "mrna_mad",
]

network_metric_columns = [
    "target_instrength_cv",
    "edge_score_dispersion",
]

correlation_rows = []

for scope_name, scope_df in [
    ("all_profiles", expression_heterogeneity),
    (
        "human_chimpanzee",
        expression_heterogeneity[
            expression_heterogeneity["Species"].isin(
                ["human", "chimpanzee"]
            )
        ],
    ),
]:
    for network_metric in network_metric_columns:
        for heterogeneity_metric in heterogeneity_columns:

            r, p = pearsonr(
                scope_df[network_metric],
                scope_df[heterogeneity_metric],
            )

            correlation_rows.append(
                {
                    "scope": scope_name,
                    "network_metric": network_metric,
                    "heterogeneity_metric": heterogeneity_metric,
                    "pearson_r": r,
                    "p_value": p,
                    "n": len(scope_df),
                }
            )

heterogeneity_correlations = pd.DataFrame(
    correlation_rows
)

display(
    heterogeneity_correlations
    .sort_values(
        ["scope", "network_metric", "pearson_r"],
        ascending=[True, True, False]
    )
    .reset_index(drop=True)
    .round(6)
)

expression_heterogeneity.to_csv(
    "results/pfc_expression_heterogeneity_mirdb.csv",
    index=False
)

heterogeneity_correlations.to_csv(
    "results/pfc_network_heterogeneity_correlations_mirdb.csv",
    index=False
)
```

18) Is expression heterogenity declining in humans but not in chimps with age?

```python
# Test whether expression heterogeneity changes differently with age in humans and chimpanzees
# Uses the same species × age model and Freedman–Lane permutation test as above.

heterogeneity_hc = (
    expression_heterogeneity[
        expression_heterogeneity["Species"].isin(
            ["human", "chimpanzee"]
        )
    ]
    .merge(
        network_features[
            [
                "profile_id",
                "log2_macaque_equivalent_age_plus_1",
            ]
        ],
        on="profile_id",
        how="left",
        validate="one_to_one",
    )
)

heterogeneity_metrics = [
    "mirna_seed_sd",
    "mirna_seed_mad",
    "mrna_sd",
    "mrna_mad",
]

species_indicator_heterogeneity = (
    heterogeneity_hc["Species"] == "human"
).astype(int).to_numpy()

aligned_age_heterogeneity = heterogeneity_hc[
    "log2_macaque_equivalent_age_plus_1"
].to_numpy()

heterogeneity_test_rows = []
heterogeneity_rng = np.random.default_rng(29356)

for metric in heterogeneity_metrics:

    result = freedman_lane_interaction_test(
        y=heterogeneity_hc[metric].to_numpy(),
        species_indicator=species_indicator_heterogeneity,
        age=aligned_age_heterogeneity,
        n_permutations=5000,
        rng=heterogeneity_rng,
    )

    result["metric"] = metric
    heterogeneity_test_rows.append(result)

heterogeneity_interaction_results = pd.DataFrame(
    heterogeneity_test_rows
)

heterogeneity_interaction_results["fdr_bh"] = (
    benjamini_hochberg(
        heterogeneity_interaction_results["permutation_p"]
    )
)

heterogeneity_interaction_results = (
    heterogeneity_interaction_results
    .sort_values("permutation_p")
    .reset_index(drop=True)
)

display(
    heterogeneity_interaction_results[
        [
            "metric",
            "chimpanzee_slope",
            "human_slope",
            "human_minus_chimpanzee_slope",
            "interaction_t",
            "analytic_p",
            "permutation_p",
            "fdr_bh",
            "r_squared",
            "n",
        ]
    ].round(6)
)

heterogeneity_interaction_results.to_csv(
    "results/pfc_expression_heterogeneity_species_age_tests.csv",
    index=False
)
```

```python
import matplotlib.pyplot as plt

# Plot the mRNA SD result from the heterogeneity analysis above.
plot_df = (
    heterogeneity_hc[
        [
            "Species",
            "log2_macaque_equivalent_age_plus_1",
            "mrna_sd",
        ]
    ]
    .dropna()
    .copy()
)

mrna_result = (
    heterogeneity_interaction_results
    .set_index("metric")
    .loc["mrna_sd"]
)

fig, ax = plt.subplots(figsize=(8, 5.5))

for species in ["chimpanzee", "human"]:
    species_df = (
        plot_df[plot_df["Species"] == species]
        .sort_values("log2_macaque_equivalent_age_plus_1")
    )

    x = species_df["log2_macaque_equivalent_age_plus_1"].to_numpy(dtype=float)
    y = species_df["mrna_sd"].to_numpy(dtype=float)

    ax.scatter(
        x,
        y,
        s=55,
        label=f"{species.capitalize()} observations",
    )

    slope, intercept = np.polyfit(x, y, 1)
    x_fit = np.linspace(x.min(), x.max(), 100)
    y_fit = intercept + slope * x_fit

    ax.plot(
        x_fit,
        y_fit,
        linewidth=2,
        label=f"{species.capitalize()} fit (slope = {slope:.4f})",
    )

ax.set_xlabel("log2(macaque-equivalent age + 1)")
ax.set_ylabel("mRNA expression SD")
ax.set_title("mRNA expression heterogeneity across development")
ax.legend(frameon=False)

annotation = (
    f"Human - chimp slope difference = "
    f"{mrna_result['human_minus_chimpanzee_slope']:.4f}\n"
    f"Freedman-Lane p = {mrna_result['permutation_p']:.4f}; "
    f"BH FDR = {mrna_result['fdr_bh']:.4f}"
)

ax.text(
    0.02,
    0.02,
    annotation,
    transform=ax.transAxes,
    va="bottom",
    ha="left",
)

fig.tight_layout()
fig.savefig(
    "results/pfc_mrna_sd_human_chimpanzee_trajectory.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```
